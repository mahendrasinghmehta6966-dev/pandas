```python
# Jab aap df.groupby() lagate hain, toh Pandas 3 steps mein kaam karta hai (Split - Apply - Combine):

#Split: Pehle woh categorical column (jaise Category ya Store) ke hisaab se data ko alag-alag groups mein tod deta hai.

#Apply: Phir har group ke numeric column (jaise Sales) par koi calculation (sum, mean, max) lagata hai.

#Combine: Aakhir mein sabhi results ko jod kar ek naya table/series dikha deta hai.
```

```python
import pandas as pd

data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 20, 8, 14, 16, 25]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
cat=df.groupby('Category')
cat
for i,v in cat:
    print(i)
    print(v)
```

```python
cat=df.groupby('Category')['Sales'].sum()
cat
```

```python
df.groupby('Store')['Sales'].sum()
```

```python
df.groupby(['Category','Store'])['Sales'].sum()
```
