```python
import numpy as np
import pandas as pd
```

```python
#merging 2 data frames
```

```python
#dataframe1
employees=pd.DataFrame({
    'employees_id':[1,2,3,4,5],
    'name':['bablu','raju','urmi','abhishek','hema'],
    'department':['hr','IT','finance','IT','hr']
})

#dataframe2
salaries=pd.DataFrame({
    'employees_id':[1,2,3,6,7],
    'salary':[60000,55000,80000,90000,50000],
    'bonus':[5000,6000,5500,8000,4000]
})


```

```python
employees
```

```python
salaries
```

```python
pd.merge(employees,salaries)
```

```python
#on lagana hogaa
```

```python
pd.merge(employees,salaries,on ='employees_id')
```

```python
pd.merge(employees,salaries,on ='employees_id',how='inner')
```

```python
#yyeh yeh kaam krta h employees id ke base pr common data ko merge kr detaa h
```

```python
#outer 
```

```python
pd.merge(employees,salaries,on ='employees_id',how='outer')
```

```python
#outer columns se hisab se merge karne kaa  kaam karte haii 
```

```python
#left ka use karenege pahli wali(employees ) wali table le liyee us ke baiss par data merge hogaa
```

```python
pd.merge(employees,salaries,on='employees_id',how='left')
```

```python
pd.merge(employees,salaries,on='employees_id',how='right')
```
